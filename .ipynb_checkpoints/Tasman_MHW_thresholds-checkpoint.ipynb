```python
# === Setup ===
base = '/g/data/e14/rmh561/access-om2/archive/025deg_jra55_iaf_cycle6_online_mlt/'
file_pattern = base + "output*/ocean/ocean_daily.nc"
file_list = sorted(glob.glob(file_pattern))

if not file_list:
    raise FileNotFoundError("No matching ocean_daily.nc files found!")

output_dir = Path("/scratch/m35/nm5072/TAS_thresholds/")
output_dir.mkdir(parents=True, exist_ok=True)

zarr_path = output_dir / "TAS_MLD_ready.zarr"
```

```python
# === Step 1: Preprocess and save to Zarr ===
print("🔄 Preprocessing and saving intermediate file to Zarr...")

ds = xr.open_mfdataset(
    file_list,
    combine="nested",
    concat_dim="time",
    compat="override",
    data_vars="minimal",
    coords="all",
    parallel=True,
    drop_variables=[
        var for var in xr.open_dataset(file_list[0]).data_vars if var != "temp_in_mld"
    ],
    chunks={},  # no chunking at load
)

# Tasman Sea region (Kajtar et al. 2022)
sreg = [135 - 360, 175 - 360, -60, -20]  # [-225, -185, -60, -20]

# Subset data to region and time
temp_mld = ds["temp_in_mld"].sel(
    xt_ocean=slice(sreg[0], sreg[1]),
    yt_ocean=slice(sreg[2], sreg[3]),
    time=slice("1988-01-01", "2018-12-31")
)

# Convert to degrees Celsius and rename
temp = (temp_mld / 1035).rename("temp")

# Wrap in dataset and chunk
ds = temp.to_dataset()
ds = ds.chunk({"time": -1, "yt_ocean": 110, "xt_ocean": 82})

# Save to Zarr
ds.to_zarr(zarr_path, mode="w")
print(f"✅ Saved intermediate Zarr file to: {zarr_path}")
```

```python
# === Calculate climatology ===
Seas = MHW.smoothedClima_mhw(ds)
Seas.to_netcdf(output_dir / "om2_025_MLT_clim.nc")
```

```python
# === Calculate threshold ===
Thresh = MHW.smoothedThresh_mhw(ds)
Thresh.to_netcdf(output_dir / "om2_025_MLT_thresh.nc")

print("✅ Done.")
```

```python
file_list = sorted(glob.glob(file_pattern))

if not file_list:
    raise FileNotFoundError("No matching ocean_daily.nc files found!")

# Open a single file to determine available variables
with xr.open_dataset(file_list[0]) as ds_sample:
    variables_to_drop = [var for var in ds_sample.data_vars if var != "temp_in_mld"]


# Open dataset without coordinates being forced to minimal
ds = xr.open_mfdataset(
    file_list,
    combine="nested",
    concat_dim="time",
    compat="override",
    data_vars="minimal",
    coords="all",
    parallel=True,
    drop_variables=variables_to_drop,
    chunks={},
    decode_times=True  # Prevent automatic decoding (we will fix manually)
)

# # Extract time from one file
# with xr.open_dataset(file_list[0]) as ds_sample:
#     correct_time = ds_sample["time"]

# # Generate full time array by concatenating time from all files
# time_values = np.concatenate([
#     xr.open_dataset(f)["time"].values for f in file_list
# ])

# # Assign correct time values
# ds = ds.assign_coords(time=("time", time_values))

# # Convert time to datetime64 if needed
# ds["time"] = pd.to_datetime(ds["time"])


# Rechunk AFTER loading to force time = -1
ds = ds.chunk({"time": -1,"yt_ocean": 108, "xt_ocean": 120})

# Convert temp_in_mld to degrees Celsius
rho0 = 1035  # kg/m^3
temp_mld = ds["temp_in_mld"] / rho0
```

### create a figure to check

```python
ts = xr.open_dataset(file_list[-1]) # load in one timestep to plot - our main dataset is not chunked for this
fig = plt.figure(figsize=(12, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

#plot the data
(ts.temp_in_mld.isel(time=-1)/rho0).plot(ax=ax, cmap=cmo.thermal, vmin=0, vmax = 30)

# Add features to the map
ax.coastlines(resolution='110m', linewidth=1)
gl = ax.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                  linewidth=0.5, color='gray', alpha=0.7, linestyle='--')
gl.top_labels = False
gl.right_labels = False

ax.add_feature(cfeature.LAND, facecolor='lightgray')
```
